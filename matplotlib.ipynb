```python
df = pd.read_csv('errorbar_dataset.csv')
df.head()
```

```python
df['Err_max'] = df['Exactitud máxima'] - df['Exactitud promedio']
df['Err_min'] = df['Exactitud promedio'] - df['Exactitud mínima']
df.head()
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
x = df['Modelo']
y = df['Exactitud promedio']
errores = [df['Err_min'], df['Err_max']]
fig, ax = plt.subplots()
ax.errorbar(x, y, yerr=errores, fmt='o--', ecolor='red', capsize=5);
```

```python
x = df['Modelo']
y = df['Exactitud promedio']
errores = [df['Err_min'], df['Err_max']]
fig, ax = plt.subplots()
marcadores, caps, barras = ax.errorbar(x, y, yerr=errores, fmt='o--', ecolor='red', capsize=5)
for cap in caps:
    cap.set_color('blue')
    cap.set_linewidth(2)
    cap.set_alpha(0.75)
for barra in barras:
    barra.set_color('blue')
    barra.set_linewidth(2)
    barra.set_alpha(0.75)
ax.set_xlabel('Modelo')
ax.set_ylabel('Exactitud promedio $\pm$ max/min')
ax.set_title('Comparación de modelos');
```

```python
data = pd.read_csv('fill_between_dataset.csv')
data.head()
```

```python
temps = data['Temp. predicha (°C)']
horas = data['Hora']
fig, ax = plt.subplots()
ax.plot(horas, temps)
ax.set_xlabel('Hora')
ax.set_ylabel('Temp. predicha (°C)');
```

```python
ci = data['CI 95% (°C)']
fig, ax = plt.subplots()
ax.errorbar(horas, temps, yerr=ci, fmt='o-', ecolor='lightblue', capsize=5)
ax.set_xlabel('Hora')
ax.set_ylabel('Temp. predicha (°C)')
ax.set_title('Predicción de temperatura con intervalos de confianza');   
```

```python
temps_mins = temps - ci
temps_maxs = temps + ci
fig, ax = plt.subplots()
ax.plot(horas, temps, 'o-', label='Temp. predicha')
ax.plot(horas, temps_mins, 'r--', label='Límite inferior')
ax.plot(horas, temps_maxs, 'r--', label='Límite superior')
ax.fill_between(horas, temps_mins, temps_maxs, color='lightblue', alpha=0.5, label='CI 95%')
ax.set_xlabel('Hora')
ax.set_ylabel('Temp. predicha (°C)')
ax.set_title('Predicción de temperatura con intervalos de confianza');
ax.legend()
```
