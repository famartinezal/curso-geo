```python
x = np.linspace(-2, 2, 100)
y1 = 5*x**2
y2 = 5*x**3
y3 = 5*x**4

fig, ax = plt.subplots()
ax.plot(x, y1, label='$5x^2$')
ax.plot(x, y2, label='$5x^3$')
ax.plot(x, y3, label='$5x^4$')
ax.set_title('Funciones polinómicas')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, which='both', linestyle='--', linewidth=0.5)
ax.legend();
```

```python
# dos graficos a la izquierda y uno a la derecha
fig, ax = plt.subplot_mosaic([['upleft', 'right'], 
                              ['lowleft', 'rigth']],
                              layout='constrained')
ax['upleft'].plot(x, x**2)
ax['lowleft'].plot(x, x**3)
ax['right'].plot(x, x**4)
ax['upleft'].set_title('Cuadrática')
ax['lowleft'].set_title('Cúbica')
ax['right'].set_title('Cuarta potencia')
```

```python
mosaico = """
ABD
CCD
"""
fig, ax = plt.subplot_mosaic(mosaico, layout='constrained')
ax['A'].plot(x, x**2)
ax['B'].plot(x, x**3)
ax['C'].plot(x, x**4)
ax['D'].plot(x, x**5)
ax['A'].set_title('Cuadrática')
ax['B'].set_title('Cúbica')
ax['C'].set_title('Cuarta potencia')
ax['D'].set_title('Quinta potencia')
```

```python
x = np.linspace(-4 * np.pi, 4 * np.pi, 200)
y = 2 * np.cos(x) * np.exp(-0.1 * x)
fig, ax = plt.subplots()
ax.plot(x, y)
ax.axis('equal')
ax.annotate('Pico', xy=(0, 2), xytext=(1, 3),
            arrowprops=dict(facecolor='black', shrink=0.05),
            fontsize=12, color='purple')
ax.annotate('Valle', xy=(2 * np.pi, -1.5), xytext=(2 * np.pi + 1, -1),
            arrowprops=dict(facecolor='black'),
            fontsize=12, color='orange')
```

```python
fig, ax = plt.subplots()
ax.matshow(matriz, cmap='coolwarm')
fig.colorbar(ax.matshow(matriz, cmap='coolwarm'), ax=ax, orientation='vertical', label='Valor')
labels = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']
ax.set_xticks(np.arange(len(labels)))
ax.set_xticklabels(labels)
ax.set_yticks(np.arange(len(labels)))
ax.set_yticklabels(labels)
```
