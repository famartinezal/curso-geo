```python
# Listas
route = [(35.6895, 139.6917), (34.0522, -118.2437), (51.5074, -0.1278)]
print(f'Route: {route}')
elevation = [120.5, 89.3, 45.0]
print(f'Elevation: {elevation}')
cities = ['Tokyo', 'Los Angeles', 'London']
print(f'Cities: {cities}')
```

```python
# borra los repetidos de la lista original
country_codes = ['JP', 'US', 'GB', 'FR', 'JP']
print(f'Original country codes: {country_codes}')
unique_country_codes = set(country_codes)
print(f'Unique country codes: {unique_country_codes}')
```

```python
# Dos áreas de estudio diferentes con sus especies observadas
area_a_species = {"roble", "pino", "arce", "abedul"}
area_b_species = {"pino", "abedul", "cedro", "abeto"}
# Especies en común entre las dos áreas
common_species = area_a_species.intersection(area_b_species)
print(f'Especies comunes entre Área A y Área B: {common_species}')  
# Especies únicas de cada área
unique_to_area_a = area_a_species.difference(area_b_species)
unique_to_area_b = area_b_species.difference(area_a_species)
print(f'Especies únicas de Área A: {unique_to_area_a}')
print(f'Especies únicas de Área B: {unique_to_area_b}')
# Todas las especies observadas en ambas áreas
all_species = area_a_species.union(area_b_species)
print(f'Todas las especies observadas en ambas áreas: {all_species}')
```
