```python
import timeit 
# listas vs arreglos
rango = [10**2, 10**3, 10**4, 10**5, 10**6]
for n in rango:
    lista = list(range(n))
    arreglo = np.arange(n)
    
    tiempo_lista = timeit.timeit('sum(lista)', globals=globals(), number=1000)
    tiempo_arreglo = timeit.timeit('np.sum(arreglo)', globals=globals(), number=1000)
    
    print(f"n={n}: Lista: {tiempo_lista:.6f} segundos, Arreglo: {tiempo_arreglo:.6f} segundos")
    # cuantas veces es más rápido el arreglo que la lista
    if tiempo_arreglo > 0:
        print(f"El arreglo es {tiempo_lista/tiempo_arreglo:.2f} veces más rápido que la lista") 
```

```python
def inicializar_centroides(datos, k, seed=None):
    if seed is not None:
        np.random.seed(seed) # Establece la semilla para reproducibilidad
    indices = np.random.choice(datos.shape[0], size=k, replace=False) # Selecciona k índices aleatorios sin reemplazo
    centroides = datos[indices] # Obtiene los centroides seleccionados
    return centroides
centroides_iniciales = inicializar_centroides(data_scaled, 3, 14) # Inicializa 3 centroides con una semilla para reproducibilidad
print(centroides_iniciales)
```

```python
def calcular_distancias(datos, centroides):
    k = centroides.shape[0] # Número de centroides
    N = datos.shape[0] # Número de datos
    distancias = np.zeros((N, k)) # Matriz para almacenar las distancias
    for i, centroide in enumerate(centroides) :
        distancias[:, i] = np.linalg.norm(datos - centroide, axis=1) # Calcula la distancia euclidiana entre cada dato y el centroide i
        
    return distancias
# prueba de la función calcular_distancias
distancias = calcular_distancias(data_scaled, centroides_iniciales)
distancias
```

```python
def clasificar(dist):
    return np.argmin(dist, axis=1) # Devuelve el índice del centroide más cercano para cada dato
# prueba de la función clasificar
y = clasificar(distancias)
idx = 7
print(f"El dato en la posición {idx} pertenece al cluster {y[idx]}")
```

```python
def actualizar_cnetroides(datos, cats):
    n_cats = cats.max() + 1 # Número de clusters (categorías)
    centroides = []
    
    for i in range(n_cats):
        cluster = datos[cats == i,:] # Datos que pertenecen al cluster i
        cnetroide = np.mean(cluster, axis=0) # Calcula el nuevo centroide como la media de los datos del cluster
        centroides.append(cnetroide) # Agrega el nuevo centroide a la lista
    return np.array(centroides) # Devuelve los nuevos centroides como un arreglo

centroides_nuevos = actualizar_cnetroides(data_scaled, y)
print(centroides_nuevos)
```
